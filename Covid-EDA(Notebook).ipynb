# Summary
### 1. There are total 5770 rows and 17 columns
### 2. There are no duplicate values
### 3. There are no row or columns that contain any null values
### 4. start_date and end_date format is object which should be in datetime format
### 5. country and region should be convert into category datatype
### 6. year, date and week should be converted into int8 datatype so that less space will be occupied
### 7. To cross check if any error exists in  calculation
###              a. (covid_deaths + non_covid_deaths = total_deaths),
###              b. (expected_deaths + excess_deaths = total_deaths) and
###              c. (end_date - start_date =days)

## Changing format from object to datetime

```python
df['start_date']=pd.to_datetime(df['start_date'],format='%d-%m-%Y')
df['end_date']=pd.to_datetime(df['end_date'],format='%d-%m-%Y')
```

```python
df.info()
```

## Correcting and optimizing the data types

```python
df['country']=df['country'].astype('category')
df['region']=df['region'].astype('category')

df['days']=df['days'].astype('int8')
df['year']=df['year'].astype('int16')
df['week']=df['week'].astype('int8')
```

```python
df.info()
```

## Cheaking for calculation errors

```python
breakdown_errors = df[abs((df['covid_deaths'] + df['non_covid_deaths']) - df['total_deaths']) > 0.1]
excess_errors = df[abs((df['expected_deaths'] + df['excess_deaths']) - df['total_deaths']) > 0.1]

calculated_days = (df['end_date'] - df['start_date']).dt.days +1
date_errors = df[calculated_days != df['days']]

print(f"Death Breakdown errors: {len(breakdown_errors)} rows")
print(f"Excess Death errors: {len(excess_errors)} rows")
print(f"Date errors: {len(date_errors)} rows")
```

# Data Visualization

```python
tline_df = df.groupby('end_date')[['total_deaths', 'expected_deaths']].sum().reset_index()
plt.figure(figsize=(12, 8))
sns.lineplot(data=tline_df, x='end_date', y='total_deaths', label='Total Deaths', color='red')
sns.lineplot(data=tline_df, x='end_date', y='expected_deaths', label='Expected Deaths', color='royalblue', linestyle='--')
plt.title('Total vs Expected Deaths',fontsize=20,fontweight='bold')
plt.xlabel('Date',fontsize=14)
plt.ylabel('Count',fontsize=14)
plt.legend(fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
top_country=df.groupby('country',observed=True)['excess_deaths'].sum().sort_values(ascending=False).head(5).index
plt.figure(figsize=(16,6))
for c in top_country:
    sub=df[df['country']==c]
    plt.plot(sub['end_date'],sub['excess_deaths_per_100k'],label=c)
plt.title('Excess death per 100k for top 5 country',fontsize=16,fontweight='bold')
plt.xlabel('Date',fontsize=14)
plt.ylabel('Excess deaths per 100k',fontsize=14)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
tcountry=df.groupby('country',observed=True)['excess_deaths_per_100k'].sum().sort_values(ascending=False).head(15)
tcountry.index = tcountry.index.astype(str)
plt.figure(figsize=(14,6))
sns.barplot(x=tcountry.values, y=tcountry.index, color='crimson',width=0.5)
plt.title('Top 15 Countries by Total Excess Deaths per 100k',fontsize=16,fontweight='bold')
plt.xlabel('Excess Deaths per 100k',fontsize=14)
plt.ylabel('Country',fontsize=14)
plt.tight_layout()
plt.show()
```

```python
tdeath=df.groupby('end_date',observed=True)[['covid_deaths','non_covid_deaths']].sum().reset_index()
plt.figure(figsize=(12,5))
sns.lineplot(data=tdeath, x='end_date', y='covid_deaths',label='Covid Deaths',color='red',linestyle='--')
sns.lineplot(data=tdeath, x='end_date', y='non_covid_deaths',label='Non Covid Deaths',color='lightgreen')
plt.title('Covid vs Non-Covid Deaths over time',fontsize=20,fontweight='bold')
plt.xlabel('Dates',fontsize=16)
plt.ylabel('Deaths',fontsize=16)
plt.tight_layout()
plt.show()
```

```python
cols = ['population', 'excess_deaths_per_100k', 'covid_deaths_per_100k', 'excess_deaths_pct_change']
plt.figure(figsize=(8, 6))
sns.heatmap(df[cols].corr(), annot=True, cmap='viridis')
plt.title('Correlation Heatmap',fontsize=16,fontweight='bold')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x='year',y='excess_deaths_pct_change')
plt.title('Excess Deaths % changes by year')
plt.ylim(-0.5, 2.5)
plt.show()
```

```python
top10 = df.groupby('country', observed=True)['excess_deaths_per_100k'].sum().sort_values(ascending=False).head(10).index
heat_data = df[df['country'].isin(top10)].copy()
heat_data['country'] = heat_data['country'].astype(str)  
heat_df = heat_data.pivot_table(index='country', columns='week', values='excess_deaths_per_100k', observed=True)
plt.figure(figsize=(12, 8))
sns.heatmap(heat_df, cmap='Reds')
plt.title('Excess Deaths per 100k: Country vs Week (Top 10 Countries)')
plt.xlabel('Week')
plt.ylabel('Country')
plt.show()
```

```python
year_df = df.groupby('year', observed=True)[['total_deaths', 'expected_deaths']].sum()
year_df.plot(kind='bar', figsize=(8, 6))
plt.title('Total vs Expected Deaths by Year')
plt.xlabel('Year',fontsize=12)
plt.ylabel('Deaths',fontsize=12)
plt.show()
```

```python
sns.pairplot(df[['population','excess_deaths_per_100k','covid_deaths_per_100k','excess_deaths_pct_change']])
plt.show()
```

```python
year_covid = df.groupby('year', observed=True)['covid_deaths'].sum()
year_covid.plot(kind='line', marker='o', figsize=(8,5))
plt.title('Covid Deaths by Year')
plt.ylabel('Deaths')
plt.show()
```

```python
country_df = df.groupby('country', observed=True)[['population', 'excess_deaths_per_100k']].sum().reset_index()
plt.figure(figsize=(8,6))
sns.scatterplot(data=country_df, x='population', y='excess_deaths_per_100k')
plt.xscale('log')
plt.title('Population vs Excess Deaths per 100k')
plt.show()
```

# Visualization Summary

### 1. Total vs Expected Deaths (Global Trend)
Total deaths tracked closely with expected deaths for most of the timeline, but showed sharp spikes above the expected baseline during major pandemic waves clearly visible as the gap between the red and blue lines.

### 2. Excess Deaths per 100k — Top 5 Countries Over Time
Different countries peaked at different times, with some  showing multiple distinct waves rather than a single spike  indicating repeated periods of health system strain rather than one-off events.

### 3. Top 15 Countries by Total Excess Deaths per 100k
Eastern European countries (Bulgaria, Lithuania, Croatia) and Latin American countries (Peru, Ecuador, Colombia) dominate the top of the ranking suggesting these regions experienced disproportionately high excess mortality relative to population.

### 4. Covid vs Non-Covid Deaths Over Time
Covid deaths and non-covid deaths both fluctuate over time, with covid deaths showing sharper spikes during wave periods, while non-covid deaths remained comparatively more stable useful for spotting periods where indirect pandemic effects  may have played a role.

### 5. Correlation Heatmap
Excess deaths per 100k and covid deaths per 100k show a strong positive correlation, confirming that covid was a major driver of excess mortality. Population shows weak/no correlation with the rate-based metrics, as expected since per-100k values are already population-normalized.

### 6. Excess Deaths % Change by Year
2020 had the most extreme outliers, reflecting the shock of the pandemic's onset. 2021 shows a denser cluster of high values, suggesting sustained elevated mortality . 2022 values are tighter around 0, indicating mortality patterns normalized as the pandemic matured.

### 7. Excess Deaths per 100k: Country vs Week Heatmap
Peak intensity (darkest cells) occurs at different weeks for different countries, confirming that pandemic waves were not globally synchronized — some countries were hit earlier, others later in the same year.

### 8. Total vs Expected Deaths by Year
Total deaths exceeded expected deaths most significantly in 2021, with 2020 and 2022 showing smaller gaps consistent with the % change findings in point 6.

### 9. Covid Deaths by Year
Covid deaths peaked in 2021 before declining, reflecting the overall trajectory of the pandemic's severity over the three year period.

### Key Takeaways
- Excess mortality was not uniform across countries Eastern Europe and Latin America were disproportionately affected.
- 2021 emerges as the most severe year overall, despite 2020 having the most extreme single outliers.
- Covid deaths correlate strongly with excess deaths, validating covid as the primary driver of the pandemic's mortality impact, though non-covid death patterns suggest some indirect effects as well.
